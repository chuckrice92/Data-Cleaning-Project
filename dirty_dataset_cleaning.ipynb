# Checking and Changing Data Types

```python
df['income'] = df['income'].replace('35k', '35000')

df['income'] = pd.to_numeric(df['income'], errors='coerce')

df['credit_score'] = pd.to_numeric(df['credit_score'], errors='coerce')

df['employment_years'] = pd.to_numeric(
    df['employment_years'], errors='coerce'
)

df['default_status'] = pd.to_numeric(
    df['default_status'], errors='coerce'
)

df['employment_years'] = pd.to_numeric(
    df['employment_years'], 
    errors="coerce"
)
```

```python
df.dtypes
```

# Data Cleaning

### Fixing Spelling Errors

```python
df['marital_status'] = df['marital_status'].replace('divorsed', 'divorced')
```

```python
df['education'] = df['education'].replace('HighSchool', 'High School')
```

## Fixing Invalid Values

### Fixing Ages

```python
df.loc[df['age'] < 0, 'age'] = pd.NA
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

### Fixing Inconsistent Values in 'product_type' and 'purchase_frequency' Columns

```python
df['product_type'] = df['product_type'].replace('3', 'Unknown')
```

```python
df['purchase_frequency'] = df['purchase_frequency'].replace('30', 'Unknown')
```

### Fixing missing Values in 'credit_score' column

```python
df['credit_score'] = df['credit_score'].fillna(df['credit_score'].median())
```

### Fixing missing Values in 'employment_years' Column

```python
df['employment_years'] = df['employment_years'].fillna(
    df['employment_years'].median()
)
```

### Eliminating Inconsistent Values in 'email' column

```python
invalid_email = ~df['email'].str.contains(
    r"^[^@\s]+@[^@\s]+\.[^@\s]+$",
    regex=True,
    na=False
)

df.loc[invalid_email, ['customer_id', 'email']]
```

```python
df.loc[invalid_email, 'email'] = pd.NA
```

```python
df['phone_number'] = (
    df['phone_number']
    .str.replace(r"\D", "", regex=True)
)
```

```python
df['phone_number']
```

### Fixing Invalid Values in 'Zip_code' Column

```python
df['zip_code'] = df['zip_code'].str.replace('"', '', regex=False)
```

```python
df['zip_code'] = df['zip_code'].str.zfill(5)
```

```python
df['zip_code']
```

### Fixing Invalid Values in 'avg_monthly_spend' Column

```python
df[df['avg_monthly_spend'] < 0]
```

```python
df.loc[df['avg_monthly_spend'] < 0, 'avg_monthly_spend'] = pd.NA
```

```python
df['avg_monthly_spend'] = df['avg_monthly_spend'].fillna(
    df['avg_monthly_spend'].median()
)
```

```python
df['avg_monthly_spend']
```

## Fixing Date Columns

```python
df['last_payment_date'] = pd.to_datetime(df['last_payment_date'], errors='coerce')
```

```python
df['last_login'] = pd.to_datetime(df['last_login'], errors='coerce')
```

```python
df['account_created'] = pd.to_datetime(df['account_created'], errors='coerce')
```

```python
df.dtypes
```

```python
df.isna().sum()
```

## Cleaning Categorical Columns

```python
categorical_columns = [
    'education',
    'marital_status',
    'state',
    'product_type'
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```
